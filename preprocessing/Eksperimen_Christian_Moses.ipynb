```python
#Ubah tipe data tanggal
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df.info()
```

```python
#Cek missing value
print("Missing value per kolom:")
print(df.isnull().sum())

# Hapus baris tanpa Customer ID
df = df.dropna(subset=["Customer ID"])
```

```python
#Buang invoice yang melakukan cancel (diawali huruf "C")
mask_cancel = df["Invoice"].astype(str).str.startswith("C")
print("Jumlah transaksi cancel:", mask_cancel.sum())
df = df[~mask_cancel]
```

```python
#Buang duplikat
print("Jumlah duplikat sebelum dihapus:", df.duplicated().sum())
df = df.drop_duplicates(keep="first")
```

```python
#Hitung total belanja per baris
for col in ["Quantity", "Price"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["Quantity", "Price"])
df["Total Amount"] = df["Quantity"] * df["Price"]
df = df[df["Total Amount"] > 0]

print("\nShape setelah cleaning:", df.shape)
```
